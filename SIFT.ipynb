```python
import cv2

# Read the two input images
img1 = cv2.imread(r"C:\Users\Rujuta\Pictures\Screenshots\box.png")
img2 = cv2.imread(r"C:\Users\Rujuta\Pictures\Screenshots\box-in-scene.png")

# Check whether images are loaded
if img1 is None or img2 is None:
    print("Error: Could not load images.")
    exit()

# Convert images to grayscale
gray1 = cv2.cvtColor(img1, cv2.COLOR_BGR2GRAY)
gray2 = cv2.cvtColor(img2, cv2.COLOR_BGR2GRAY)

# Create SIFT detector
sift = cv2.SIFT_create()

# Detect key points and compute descriptors
keypoints1, descriptors1 = sift.detectAndCompute(gray1, None)
keypoints2, descriptors2 = sift.detectAndCompute(gray2, None)

# Create Brute-Force matcher
bf = cv2.BFMatcher()

# Find matches using KNN
matches = bf.knnMatch(descriptors1, descriptors2, k=2)

# Apply Lowe's ratio test
good_matches = []

for m, n in matches:
    if m.distance < 0.75 * n.distance:
        good_matches.append(m)

# Draw the good matches
result = cv2.drawMatches(
    img1, keypoints1,
    img2, keypoints2,
    good_matches, None,
    flags=cv2.DrawMatchesFlags_NOT_DRAW_SINGLE_POINTS
)

# Display results
print("Key points in Image 1:", len(keypoints1))
print("Key points in Image 2:", len(keypoints2))
print("Good matches found:", len(good_matches))

cv2.imshow("SIFT Feature Matching", result)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2

# Read the two input images
img1 = cv2.imread(r"C:\Users\Rujuta\Pictures\Screenshots\box.png")
img2 = cv2.imread(r"C:\Users\Rujuta\Pictures\Screenshots\box-in-scene.png")

# Check whether images are loaded
if img1 is None or img2 is None:
    print("Error: Could not load images.")
    exit()

# Display the original images
cv2.imshow("Image 1 - Original", img1)
cv2.imshow("Image 2 - Original", img2)

# Convert images to grayscale
gray1 = cv2.cvtColor(img1, cv2.COLOR_BGR2GRAY)
gray2 = cv2.cvtColor(img2, cv2.COLOR_BGR2GRAY)

# Create SIFT detector
sift = cv2.SIFT_create()

# Detect key points and compute descriptors
keypoints1, descriptors1 = sift.detectAndCompute(gray1, None)
keypoints2, descriptors2 = sift.detectAndCompute(gray2, None)

# Create Brute-Force matcher
bf = cv2.BFMatcher()

# Find matches using KNN
matches = bf.knnMatch(descriptors1, descriptors2, k=2)

# Apply Lowe's ratio test
good_matches = []

for m, n in matches:
    if m.distance < 0.75 * n.distance:
        good_matches.append(m)

# Draw the good matches
result = cv2.drawMatches(
    img1, keypoints1,
    img2, keypoints2,
    good_matches, None,
    flags=cv2.DrawMatchesFlags_NOT_DRAW_SINGLE_POINTS
)

# Display the SIFT matching result
cv2.imshow("SIFT Feature Matching", result)

# Display information
print("Key points in Image 1:", len(keypoints1))
print("Key points in Image 2:", len(keypoints2))
print("Good matches found:", len(good_matches))

# Wait until a key is pressed
cv2.waitKey(0)

# Close all windows
cv2.destroyAllWindows()
```
